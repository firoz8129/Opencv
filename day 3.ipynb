```python
#cropping
import cv2 as cv
img=cv.imread('cat.jpg')

imgcropped=img[100:500,200:600]
cv.imshow('org',img)
cv.imshow('crop',imgcropped)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# resizing
import cv2 as cv
img=cv.imread("cat.jpg")
print(img.shape)
#imgrgb=cv.cvtColor(img,cv.COLOR_BGR2RGB)
imgrsiz=cv.resize(img,(600,920))
imggray=cv.cvtColor(imgrsiz,cv.COLOR_BGR2GRAY)
intr_nearest=cv.resize(imggray,(600,920),interpolation=cv.INTER_NEAREST)
intr_linear=cv.resize(imggray,(600,900),interpolation=cv.INTER_LINEAR)
intr_cbic=cv.resize(imggray,(600,900),interpolation=cv.INTER_CUBIC)
intr_area=cv.resize(imggray,(600,900),interpolation=cv.INTER_AREA)
cv.imshow('org',img)
cv.imshow('resize',imgrsiz)
cv.imshow('nearst',intr_nearest)
cv.imshow('lnr',intr_linear)
cv.imshow('cubic',intr_cbic)
cv.imshow('area',intr_area)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Resize an image to half its size and crop the center portion.
import cv2 as cv

img = cv.imread("cat.jpg")

# Resize to half
h, w = img.shape[:2]
resize = cv.resize(img, (w//2, h//2))
print(resize.shape)

# Crop center
h, w = resize.shape[:2]
crop = resize[h//4:3*h//4, w//4:3*w//4]
print(crop.shape)

cv.imshow("Resized", resize)
cv.imshow("Center Crop", crop)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Resize image to 200 × 200
# Crop top-left 100 × 100
# Resize cropped part back to original image size

import cv2 as cv

img=cv.imread("cat.jpg")

resize=cv.resize(img,(200,200))

cropped= resize[0:100,0:100]

original_h,original_w=cropped.shape[:2]
back_to_original =cv.resize(cropped, ( original_w,original_h))

cv.imshow("original image",img)
cv.imshow("200 *200",resize)
cv.imshow("crop",cropped)
cv.imshow("resize crop",back_to_original)

cv.waitKey(0)
cv.destroyAllWindows()
```
