```python
import cv2 as cv
import numpy as np
```

```python
img=cv.imread("9cad3ab8-f440-413d-bb8d-f1ecab57c96e.jpg",1)
cv.imshow("my img",img)
cv.waitKey(5000)
cv.imwrite("sv img.jpg",img)
cv.destroyAllWindows()
```

```python
clr_img=cv.imread("9cad3ab8-f440-413d-bb8d-f1ecab57c96e.jpg",cv.IMREAD_COLOR)

gry_img=cv.imread("9cad3ab8-f440-413d-bb8d-f1ecab57c96e.jpg",cv.IMREAD_GRAYSCALE)

cv.imshow("clr_img",clr_img)
cv.imshow("gry_img",gry_img)

cv.waitKey(2500)

cv.imwrite("save_gry.jpg",gry_img)

cv.destroyAllWindows()
```

```python
img=cv.imread("9cad3ab8-f440-413d-bb8d-f1ecab57c96e.jpg",1)
imggray=cv.cvtColor(img,cv.COLOR_BGRA2GRAY)
cv.imshow("gray img",imggray)
cv.waitKey(5000)
cv.imwrite("sv img.jpg",imggray)
cv.destroyAllWindows()
```
