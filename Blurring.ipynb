```python
import cv2 as cv

img=cv.imread("cat.jpg")

blur=cv.GaussianBlur(img,(11,11),0)
cv.imshow("org",img)
cv.imshow("blr",blur)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv

img=cv.imread("cat.jpg",1)
imgrz=cv.resize(img,(1900,800))
imgrgb=cv.cvtColor(imgrz,cv.COLOR_BGRA2BGR)
cv.imshow("resize",imgrz)
cv.imshow("rgb",imgrgb)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
G_blur=cv.GaussianBlur(imgrgb,(15,15),0)
cv.imshow("Gousian",G_blur)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
M_blur=cv.medianBlur(imgrgb,5)

cv.imshow("median",M_blur)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
B_blur=cv.bilateralFilter(imgrgb,9,75,75)
cv.imshow("bilaterial",B_blur)
cv.waitKey(0)
cv.destroyAllWindows()
```
