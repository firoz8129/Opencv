```python
import cv2 as cv
img1=cv.imread("cat.jpg",1)
grey=cv.cvtColor(img1,cv.COLOR_BGR2GRAY)
gray1=cv.imread("cat.jpg",cv.IMREAD_GRAYSCALE)

cv.imshow("org",img1)
cv.imshow('grey',grey)
cv.imshow('gr',gray1)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv
#resize
img2=cv.imread("cat.jpg")
print(img2.shape)
imgre=cv.resize(img2,(400,600))

imgcr=img2[100:300,300:700]

cv.imshow("re",imgre)
cv.imshow('cr',imgcr)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv
img2=cv.imread("cat.jpg")
h,w=img2.shape[:2]
resize=cv.resize(img2,(h//2,w//2))

h,w=resize.shape[:2]
crprsz=resize[h//4:3*h//4,w//4:3*w//4]

original_h,original_w=crprsz.shape[:2]
back_to_original=cv.resize(crprsz,(original_h,original_w))

cv.imshow("rehalf",resize)
cv.imshow("recrp",crprsz)
cv.imshow("org back",back_to_original)
cv.waitKey(0)
cv.destroyAllWindows()
```
