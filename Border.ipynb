```python
import cv2 as cv
img=cv.imread("cat.jpg")
img_border=cv.copyMakeBorder(img,15,15,10,10,cv.BORDER_CONSTANT,None,value=0)
cv.imshow("border",img_border)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv
img=cv.imread("cat.jpg")
img_borderr=cv.copyMakeBorder(img,2000,2000,1500,500,cv.BORDER_REFLECT)
cv.imshow("border",img_borderr)
cv.waitKey(0)
cv.destroyAllWindows()
```
