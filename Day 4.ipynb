```python
import cv2 as cv

img=cv.imread("sea.jpg",cv.IMREAD_GRAYSCALE)

# 1. Simple Thresholding
_,simple=cv.threshold(img,127,255,cv.THRESH_BINARY)

# 2. Adaptive Thresholding
adaptive=cv.adaptiveThreshold(img,255,cv.ADAPTIVE_THRESH_GAUSSIAN_C,cv.THRESH_BINARY,11,2)

# 3. Otsu's Thresholding
_,otsu=cv.threshold(img,0,255,cv.THRESH_BINARY + cv.THRESH_OTSU)

# Display results
cv.imshow("Original Grayscale", img)
cv.imshow("Simple Thresholding", simple)
cv.imshow("Adaptive Thresholding", adaptive)
cv.imshow("Otsu Thresholding", otsu)

cv.waitKey(0)
cv.destroyAllWindows()
```

Thresholding

```python
import cv2 as cv

img=cv.imread("high-speed-race-car-illustration-ai-generative-free-photo.jpg",cv.IMREAD_GRAYSCALE)

#threshold value
t=127

#1 Binary threshold
_,binary=cv.threshold(img,t,255,cv.THRESH_BINARY)

#2 bnry inverted
_,binary_inv=cv.threshold(img,t,255,cv.THRESH_BINARY_INV)

#Truncated thrshld
_,trunc=cv.threshold(img,t,255,cv.THRESH_TRUNC)

#to zero thrshld
_,tozero=cv.threshold(img,t,255,cv.THRESH_TOZERO)

#tozero inv
_,tozero_inv=cv.threshold(img,t,255,cv.THRESH_TOZERO_INV)

cv.imshow("Original", img)
cv.imshow("Binary", binary)
cv.imshow("Binary Inverted", binary_inv)
cv.imshow("Truncated", trunc)
cv.imshow("To Zero", tozero)
cv.imshow("To Zero Inverted", tozero_inv)

cv.waitKey(0)
cv.destroyAllWindows()
```

Adaptive thresholding


```python
import cv2 as cv

img =cv.imread("text image.jpg",cv.IMREAD_GRAYSCALE)
t=127

_,binary=cv.threshold(img,t,255,cv.THRESH_BINARY)

mean=cv.adaptiveThreshold(img,255,cv.ADAPTIVE_THRESH_MEAN_C,cv.THRESH_BINARY,11,2)

gaussian=cv.adaptiveThreshold(img,255,cv.ADAPTIVE_THRESH_GAUSSIAN_C,cv.THRESH_BINARY,11,2)

cv.imshow("org",img)
cv.imshow("smple threshold",binary)
cv.imshow("mean thrshld",mean)
cv.imshow("gaussian",gaussian)

cv.waitKey(0)
cv.destroyAllWindows()
```
